```python
import pandas as pd
root = '/home/brakine/VMD_STL_Parallel_TimesNet_BiLSTM_POWERFORECASTING/'
Dataset = pd.read_csv(root + 'data/Yulara/dataset_Laundry_modified.csv').drop(index=0)
Dataset.head()
```

## Correlation matrix

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Calculate the correlation matrix
correlation_matrix = Dataset.corr()

# Create a heatmap of the correlation matrix
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, center=0)
plt.title('Matriz de correlación de Yulara Dataset')
plt.show()
```

## Distribution of the data

```python
n = len(Dataset)
train_n = int(n * 0.6)
valid_n = int(n * 0.2)
test_n = int(n * 0.2)

train_data = Dataset.iloc[:train_n]
valid_data = Dataset.iloc[train_n:train_n+valid_n]
test_data = Dataset.iloc[train_n+valid_n:]
```

```python
plt.figure(figsize=(10,5))
plt.plot(train_data.index, train_data["Active_Power"], label='Potencia Activa de entrenamiento (kW)')
plt.plot(valid_data.index, valid_data["Active_Power"], label='Potencia Activa de validación (kW)')
plt.plot(test_data.index, test_data["Active_Power"], label='Potencia Activa de prueba (kW)')
plt.title('Visualización de los datos ploteados')
plt.xlabel('Muestras'); plt.ylabel('Variable de salida')
plt.legend(); plt.grid(True); plt.tight_layout(); plt.show()
```
